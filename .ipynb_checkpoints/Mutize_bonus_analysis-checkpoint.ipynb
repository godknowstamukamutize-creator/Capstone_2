# EmporiUm Bonus Analysis
 
**Territories:** Miami Vue (New Jersey) & See Ellefson (New York)  
This notebook addresses the bonus questions from the Capstone 2 appendix.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
from scipy import stats

# Load all datasets
store_detail = pd.read_csv('StoreDetail.csv')
store_sales  = pd.read_csv('StoreSales.csv')
products     = pd.read_csv('Products.csv')
categories   = pd.read_csv('ProductCategories.csv')
customers    = pd.read_csv('customer_list.csv', sep='|')
customers.columns = customers.columns.str.strip()

# Shared setup
store_sales['Transaction Date'] = pd.to_datetime(store_sales['Transaction Date'])
store_sales['Month'] = store_sales['Transaction Date'].dt.to_period('M')
store_sales['Year']  = store_sales['Transaction Date'].dt.year

instore = store_sales[store_sales['Store ID'].notna() & (store_sales['Store ID'] != 0)]

t1 = store_detail[store_detail['Territory Manager'] == 'Miami Vue']
t2 = store_detail[store_detail['Territory Manager'] == 'See Ellefson']
t1_store_ids = t1['Store ID'].tolist()
t2_store_ids = t2['Store ID'].tolist()

t1_sales = instore[instore['Store ID'].isin(t1_store_ids)]
t2_sales = instore[instore['Store ID'].isin(t2_store_ids)]

print('Setup complete.')
print(f'Territory 1 (Miami Vue) stores: {t1_store_ids}')
print(f'Territory 2 (See Ellefson) stores: {t2_store_ids}')
```

---
## Filtering, Sorting & Exploration

### Q1: Create a list of all transactions from January 2024, sorted by sale amount from highest to lowest. Which transactions had a sale amount greater than $500?

```python
# Filter for January 2024 across all stores
jan_2024 = store_sales[
    (store_sales['Transaction Date'].dt.year == 2024) &
    (store_sales['Transaction Date'].dt.month == 1)
].sort_values('Sale Amount', ascending=False)

print(f'Total transactions in January 2024: {len(jan_2024)}')
print()
print('All January 2024 transactions (highest to lowest):')
print(jan_2024[['Transaction Date', 'Store ID', 'Prod Num', 'Sale Amount']].to_string())

# Transactions over $500
over_500 = jan_2024[jan_2024['Sale Amount'] > 500]
print(f'\nTransactions over $500: {len(over_500)}')
print(over_500[['Transaction Date', 'Store ID', 'Prod Num', 'Sale Amount']].to_string())
```

### Q2: Find all products whose product number begins with 10525. What category and subcategory do they belong to?

```python
# Filter products starting with 10525
prods_10525 = products[products['Prod Num'].str.startswith('10525')]

# Merge with categories to get category and subcategory names
prods_10525 = prods_10525.merge(categories, on=['CategoryID', 'SubcategoryID'], how='left')

print(f'Products starting with 10525: {len(prods_10525)}')
print()
print(prods_10525[['Prod Num', 'Product', 'Category', 'Subcategory']].to_string())
```

### Q3: Which stores in your assigned territories had at least one transaction with a sale amount over $1,000? How many such transactions did each store have?

```python
# Combine both territory sales
both_territories = pd.concat([t1_sales, t2_sales])

# Filter for transactions over $1000
over_1000 = both_territories[both_territories['Sale Amount'] > 1000]

# Count per store
over_1000_by_store = over_1000.groupby('Store ID').size().reset_index(name='Transactions Over $1000')
over_1000_by_store = over_1000_by_store.merge(store_detail[['Store ID', 'Store Location', 'Territory Manager']], on='Store ID')
over_1000_by_store = over_1000_by_store.sort_values('Transactions Over $1000', ascending=False)

print('Stores with transactions over $1,000:')
print(over_1000_by_store[['Store Location', 'Territory Manager', 'Transactions Over $1000']].to_string(index=False))
```

---
## Aggregation & NumPy

### Q4: What is the total sales revenue across all in-store transactions? What is the average transaction amount? What is the standard deviation?

```python
total_revenue = instore['Sale Amount'].sum()
avg_transaction = np.mean(instore['Sale Amount'])
std_transaction = np.std(instore['Sale Amount'])

print(f'Total in-store revenue:        ${total_revenue:,.2f}')
print(f'Average transaction amount:    ${avg_transaction:,.2f}')
print(f'Standard deviation:            ${std_transaction:,.2f}')
```

### Q5: Which stores had total revenue above the average store revenue across the full dataset?

```python
# Total revenue per store
store_revenue = instore.groupby('Store ID')['Sale Amount'].sum().reset_index()
store_revenue.columns = ['Store ID', 'Total Revenue']
store_revenue = store_revenue.merge(store_detail[['Store ID', 'Store Location', 'Territory Manager']], on='Store ID')

# Average store revenue
avg_store_revenue = np.mean(store_revenue['Total Revenue'])
print(f'Average store revenue across all stores: ${avg_store_revenue:,.2f}')
print()

# Stores above average
above_avg = store_revenue[store_revenue['Total Revenue'] > avg_store_revenue].sort_values('Total Revenue', ascending=False)
print(f'Stores above average ({len(above_avg)} stores):')
print(above_avg[['Store Location', 'Territory Manager', 'Total Revenue']].to_string(index=False))
```

### Q6: What is the total revenue for each product category across all territories, sorted highest to lowest? Which category has the most individual transactions?

```python
# Merge sales with products and categories
sales_cat = instore.merge(products[['Prod Num', 'CategoryID']], on='Prod Num', how='left')
sales_cat = sales_cat.merge(categories[['CategoryID', 'Category']].drop_duplicates(), on='CategoryID', how='left')

cat_summary = sales_cat.groupby('Category').agg(
    Total_Revenue=('Sale Amount', 'sum'),
    Transactions=('Sale Amount', 'count')
).reset_index().sort_values('Total_Revenue', ascending=False)

print('Revenue by category (highest to lowest):')
print(cat_summary.to_string(index=False))
print()
print(f"Most transactions: {cat_summary.sort_values('Transactions', ascending=False).iloc[0]['Category']}")
```

```python
# Chart: Revenue by category
fig, ax = plt.subplots(figsize=(10, 5))
ax.barh(cat_summary['Category'], cat_summary['Total_Revenue'])
ax.set_title('Total Revenue by Product Category (All Territories)')
ax.set_xlabel('Total Revenue ($)')
ax.xaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'${x:,.0f}'))
ax.invert_yaxis()
plt.tight_layout()
plt.show()
```

### Q7: Across the full dataset, what are the top 10 best-selling individual products by total revenue?

```python
# Merge sales with product names
sales_prods = instore.merge(products[['Prod Num', 'Product']], on='Prod Num', how='left')

top10_products = sales_prods.groupby('Product')['Sale Amount'].sum().reset_index()
top10_products.columns = ['Product', 'Total Revenue']
top10_products = top10_products.sort_values('Total Revenue', ascending=False).head(10)

print('Top 10 best-selling products by revenue:')
print(top10_products.to_string(index=False))
```

---
## Merging & Grouping

### Q8: For each transaction in your territories, display the transaction date, sale amount, city, state, and territory manager.

```python
# Merge territory sales with store detail
both_territories = pd.concat([t1_sales, t2_sales])
territory_detail = both_territories.merge(
    store_detail[['Store ID', 'Store Location', 'State', 'Territory Manager']],
    on='Store ID', how='left'
)

print(f'Total territory transactions: {len(territory_detail)}')
print()
print(territory_detail[['Transaction Date', 'Sale Amount', 'Store Location', 'State', 'Territory Manager']]
      .sort_values('Transaction Date')
      .head(20)
      .to_string(index=False))
```

### Q9: What is total revenue by region across the full dataset? Which region generates the most revenue?

```python
# Merge all instore sales with region info
sales_region = instore.merge(store_detail[['Store ID', 'Region']], on='Store ID', how='left')

region_revenue = sales_region.groupby('Region')['Sale Amount'].sum().reset_index()
region_revenue.columns = ['Region', 'Total Revenue']
region_revenue = region_revenue.sort_values('Total Revenue', ascending=False)

print('Revenue by region:')
print(region_revenue.to_string(index=False))
print()
print(f"Top region: {region_revenue.iloc[0]['Region']} (${region_revenue.iloc[0]['Total Revenue']:,.2f})")
```

```python
# Chart: Revenue by region
fig, ax = plt.subplots(figsize=(8, 4))
ax.bar(region_revenue['Region'], region_revenue['Total Revenue'])
ax.set_title('Total Revenue by Region')
ax.set_xlabel('Region')
ax.set_ylabel('Total Revenue ($)')
ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'${x:,.0f}'))
plt.tight_layout()
plt.show()
```

### Q10: For each product subcategory in your territories, calculate the total revenue and number of transactions. Which subcategories are driving the most revenue?

```python
# Merge territory sales → products → categories (subcategory level)
both_territories = pd.concat([t1_sales, t2_sales])
subcat_sales = both_territories.merge(products[['Prod Num', 'SubcategoryID']], on='Prod Num', how='left')
subcat_sales = subcat_sales.merge(categories[['SubcategoryID', 'Subcategory']].drop_duplicates(), on='SubcategoryID', how='left')

subcat_summary = subcat_sales.groupby('Subcategory').agg(
    Total_Revenue=('Sale Amount', 'sum'),
    Transactions=('Sale Amount', 'count')
).reset_index().sort_values('Total_Revenue', ascending=False)

print('Subcategory revenue in your territories (highest to lowest):')
print(subcat_summary.to_string(index=False))
```

### Q11: Identify all rewards members who made purchases in your territories. What is the average transaction value for rewards members vs non-rewards customers?

```python
both_territories = pd.concat([t1_sales, t2_sales]).copy()
customers['cust_id'] = customers['cust_id'].astype(str)
both_territories['RewardsID'] = both_territories['RewardsID'].astype(str)

# Tag rewards vs non-rewards
both_territories['Is Rewards Member'] = both_territories['RewardsID'].isin(customers['cust_id'])

avg_by_type = both_territories.groupby('Is Rewards Member')['Sale Amount'].mean().reset_index()
avg_by_type['Customer Type'] = avg_by_type['Is Rewards Member'].map({True: 'Rewards Member', False: 'Non-Rewards'})

print('Average transaction value by customer type:')
print(avg_by_type[['Customer Type', 'Sale Amount']].to_string(index=False))

rewards_members = both_territories[both_territories['Is Rewards Member']]
print(f'\nTotal rewards members who made purchases: {rewards_members["RewardsID"].nunique()}')
```

### Q12: Who are the top 10 rewards members by total spend across the full dataset? How many transactions does each have?

```python
# Use full instore dataset (not just territories)
rewards_sales = instore[instore['RewardsID'].notna()].copy()
rewards_sales['RewardsID'] = rewards_sales['RewardsID'].astype(str)
customers['cust_id'] = customers['cust_id'].astype(str)

top10_rewards = rewards_sales.groupby('RewardsID').agg(
    Total_Spent=('Sale Amount', 'sum'),
    Transactions=('Sale Amount', 'count')
).reset_index()

top10_rewards = top10_rewards.rename(columns={'RewardsID': 'cust_id'})
top10_rewards = top10_rewards.merge(customers[['cust_id', 'name', 'email']], on='cust_id', how='left')
top10_rewards = top10_rewards.sort_values('Total_Spent', ascending=False).head(10)

print('Top 10 rewards members by total spend (full dataset):')
print(top10_rewards[['name', 'email', 'Total_Spent', 'Transactions']].to_string(index=False))
```

## Advanced / Stretch

### Q13: Compare year-over-year revenue for your territories. Is each territory trending up or down between years?

```python
t1_yoy = t1_sales.groupby('Year')['Sale Amount'].sum().reset_index()
t1_yoy.columns = ['Year', 'Revenue']

t2_yoy = t2_sales.groupby('Year')['Sale Amount'].sum().reset_index()
t2_yoy.columns = ['Year', 'Revenue']

print('Territory 1 (Miami Vue) Year-over-Year Revenue:')
print(t1_yoy.to_string(index=False))
t1_trend = 'up' if t1_yoy['Revenue'].iloc[-1] > t1_yoy['Revenue'].iloc[0] else 'down'
print(f'Trend: {t1_trend}')

print()
print('Territory 2 (See Ellefson) Year-over-Year Revenue:')
print(t2_yoy.to_string(index=False))
t2_trend = 'up' if t2_yoy['Revenue'].iloc[-1] > t2_yoy['Revenue'].iloc[0] else 'down'
print(f'Trend: {t2_trend}')
```

```python
# Chart: Year-over-year comparison
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(t1_yoy['Year'], t1_yoy['Revenue'], marker='o', label='Miami Vue (NJ)')
ax.plot(t2_yoy['Year'], t2_yoy['Revenue'], marker='o', label='See Ellefson (NY)')
ax.set_title('Year-over-Year Revenue by Territory')
ax.set_xlabel('Year')
ax.set_ylabel('Revenue ($)')
ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'${x:,.0f}'))
ax.legend()
plt.tight_layout()
plt.show()
```

### Q14: Identify the single best month and single worst month for each territory over the full data period.

```python
t1_monthly = t1_sales.groupby('Month')['Sale Amount'].sum().reset_index()
t1_monthly.columns = ['Month', 'Revenue']

t2_monthly = t2_sales.groupby('Month')['Sale Amount'].sum().reset_index()
t2_monthly.columns = ['Month', 'Revenue']

print('Territory 1 (Miami Vue):')
print(f"  Best month:  {t1_monthly.loc[t1_monthly['Revenue'].idxmax(), 'Month']} — ${t1_monthly['Revenue'].max():,.2f}")
print(f"  Worst month: {t1_monthly.loc[t1_monthly['Revenue'].idxmin(), 'Month']} — ${t1_monthly['Revenue'].min():,.2f}")

print()
print('Territory 2 (See Ellefson):')
print(f"  Best month:  {t2_monthly.loc[t2_monthly['Revenue'].idxmax(), 'Month']} — ${t2_monthly['Revenue'].max():,.2f}")
print(f"  Worst month: {t2_monthly.loc[t2_monthly['Revenue'].idxmin(), 'Month']} — ${t2_monthly['Revenue'].min():,.2f}")
```

### Q15: Using monthly revenue totals, fit a linear regression model and plot the trend line alongside the monthly revenue chart. Is your territory trending up or down overall?

```python
# Convert Month period to numeric index for regression
t1_monthly['Month_num'] = range(len(t1_monthly))
t2_monthly['Month_num'] = range(len(t2_monthly))

# Fit linear regression for both territories
t1_slope, t1_intercept, _, _, _ = stats.linregress(t1_monthly['Month_num'], t1_monthly['Revenue'])
t2_slope, t2_intercept, _, _, _ = stats.linregress(t2_monthly['Month_num'], t2_monthly['Revenue'])

t1_trendline = t1_slope * t1_monthly['Month_num'] + t1_intercept
t2_trendline = t2_slope * t2_monthly['Month_num'] + t2_intercept

# Plot
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for ax, monthly, trendline, slope, title in zip(
    axes,
    [t1_monthly, t2_monthly],
    [t1_trendline, t2_trendline],
    [t1_slope, t2_slope],
    ['Territory 1: Miami Vue (NJ)', 'Territory 2: See Ellefson (NY)']
):
    ax.plot(monthly['Month'].astype(str), monthly['Revenue'], marker='o', label='Monthly Revenue', alpha=0.7)
    ax.plot(monthly['Month'].astype(str), trendline, linestyle='--', color='red',
            label=f'Trend ({"↑ up" if slope > 0 else "↓ down"})')
    ax.set_title(title)
    ax.set_xlabel('Month')
    ax.set_ylabel('Revenue ($)')
    ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'${x:,.0f}'))
    ax.tick_params(axis='x', rotation=45)
    ax.legend()

plt.tight_layout()
plt.show()

print(f'Territory 1 trend: {"UPWARD" if t1_slope > 0 else "DOWNWARD"} (slope = ${t1_slope:,.2f}/month)')
print(f'Territory 2 trend: {"UPWARD" if t2_slope > 0 else "DOWNWARD"} (slope = ${t2_slope:,.2f}/month)')
```
